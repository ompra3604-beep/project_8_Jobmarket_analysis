<h2><span style="color:lightblue"><b>Stage 2: Clean data and engineer features</b></span></h2>
Missing prices are preserved as missing—not converted to zero. Duplicate links are removed where possible.
Before conducting the analysis, we need to clean the data.

```python
jobs = df.copy()
jobs["title"] = jobs["title"].astype("string").str.strip()
jobs["link"] = jobs["link"].astype("string").str.strip()
jobs["country"] = jobs["country"].astype("string").str.strip()
jobs = jobs[jobs["title"].notna() & jobs["title"].ne("")].copy()

has_link = jobs["link"].notna() & jobs["link"].ne("")
with_links = jobs[has_link].drop_duplicates(subset=["link"], keep="first")
without_links = jobs[~has_link].drop_duplicates()
jobs = pd.concat([with_links, without_links], ignore_index=True)

jobs["published_date"] = pd.to_datetime(jobs["published_date"], errors="coerce", utc=True)
for col in ["hourly_low", "hourly_high", "budget"]:
    jobs[col] = pd.to_numeric(jobs[col], errors="coerce")
    jobs.loc[jobs[col] < 0, col] = np.nan

jobs["country"] = jobs["country"].fillna("Unknown").replace("", "Unknown")
jobs["is_hourly"] = jobs["is_hourly"].fillna(False).astype(bool)
jobs["pricing_type"] = np.where(jobs["is_hourly"], "Hourly", "Fixed budget")
jobs["hourly_midpoint"] = jobs[["hourly_low", "hourly_high"]].mean(axis=1)
jobs["hourly_midpoint"] = jobs["hourly_midpoint"].fillna(jobs["hourly_low"]).fillna(jobs["hourly_high"])
jobs["month"] = jobs["published_date"].dt.to_period("M").dt.to_timestamp()
jobs["year"] = jobs["published_date"].dt.year
jobs["title_clean"] = (jobs["title"].str.lower()
    .str.replace(r"[^a-z0-9+#.\s]", " ", regex=True)
    .str.replace(r"\s+", " ", regex=True).str.strip())

print(f"Original rows: {len(df):,}")
print(f"After cleaning/deduplication: {len(jobs):,}")
print(f"Date range: {jobs['published_date'].min()} → {jobs['published_date'].max()}")
display(jobs.head())
```

<h2><span style="color:lightblue"><b>Stage 4: Job Category Classification Using NLP</b></span></h2>

Categories are inferred from title keywords using transparent rules. Review and extend the rules for your use case; titles may fit multiple categories.

The dataset contains job titles rather than a dedicated job-category column. We can use Natural Language Processing to organize postings into meaningful categories.

- Potential categories include:
- Data Science and Machine Learning
- Data Analytics and Business Intelligence
- Software Development
- Web Development
- Mobile App Development
- Digital Marketing and SEO
- Graphic Design and UI/UX
- Content Writing and Translation
- Finance and Accounting
- Project Management
- Other / Unclassified

We can initially create rule-based categories and then improve the system using text clustering.

```python
CATEGORY_RULES = {
    "Data Science & Machine Learning": ["data scientist","machine learning","deep learning","artificial intelligence","nlp","computer vision","llm","generative ai"],
    "Data Analytics & BI": ["data analyst","data analysis","power bi","tableau","business intelligence","sql analyst","excel dashboard","analytics"],
    "Software Development": ["software engineer","software developer","python developer","java developer","backend developer","full stack","api development","programming"],
    "Web Development": ["web developer","wordpress","shopify","frontend developer","front end","webflow","website development"],
    "Mobile Development": ["android developer","ios developer","flutter","react native","mobile app"],
    "Design & Creative": ["graphic design","graphic designer","ui ux","ui/ux","logo design","figma","illustration"],
    "Digital Marketing": ["digital marketing","seo","sem","social media","google ads","facebook ads","content marketing"],
    "Writing & Translation": ["content writer","copywriter","technical writer","translation","proofreading","blog writer"],
    "Finance & Accounting": ["accountant","bookkeeping","financial analyst","financial modeling","quickbooks"],
    "Customer Support & Admin": ["virtual assistant","customer support","customer service","data entry","administrative assistant"],
    "Video & Animation": ["video editor","video editing","animation","motion graphics","youtube video"]
}
def categorize_title(title):
    t = str(title).lower()
    for category, keywords in CATEGORY_RULES.items():
        if any(k in t for k in keywords):
            return category
    return "Other / Unclassified"

jobs["job_category"] = jobs["title"].map(categorize_title)
category_counts = jobs["job_category"].value_counts()
display(category_counts.rename_axis("job_category").to_frame("postings"))
top_cat = category_counts.sort_values().tail(15)
fig = px.bar(x=top_cat.values, y=top_cat.index, orientation="h",
             color=top_cat.values, color_continuous_scale="Turbo",
             title="Job postings by inferred category (top 15)",
             labels={"x":"Postings","y":"Inferred category","color":"Postings"})
fig.update_layout(height=650)
fig.show()
```

<h2><span style="color:lightblue"><b>Insights</b></span></h2>

This is a baseline classification method. It should be validated against actual titles because some jobs may belong to multiple categories, and keyword rules can miss unfamiliar terminology.

Only records with valid target values should be used for training. In addition, the job title may contain information about the expected price, so we should evaluate whether text features improve performance.

These models would estimate advertised project prices, not necessarily employee salaries.

```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Prepare the text used for matching
jobs["recommendation_text"] = (
    jobs["title"].fillna("")
    + " "
    + jobs["job_category"].fillna("")
)

# Fit TF-IDF on job descriptions available in this dataset
vectorizer = TfidfVectorizer(
    stop_words="english",
    ngram_range=(1, 2),
    max_features=20000
)

job_vectors = vectorizer.fit_transform(
    jobs["recommendation_text"]
)


def recommend_jobs(user_skills, top_n=10):
    """
    Recommend jobs based on similarity between
    user interests and job titles/categories.
    """

    user_vector = vectorizer.transform([user_skills])

    scores = cosine_similarity(
        user_vector,
        job_vectors
    ).ravel()

    results = jobs.copy()
    results["relevance_score"] = scores

    results = results[
        results["relevance_score"] > 0
    ]

    results = results.sort_values(
        "relevance_score",
        ascending=False
    )

    columns = [
        "title",
        "job_category",
        "country",
        "is_hourly",
        "hourly_low",
        "hourly_high",
        "budget",
        "published_date",
        "link",
        "relevance_score"
    ]

    return results[columns].head(top_n)


# Example recommendation
recommendations = recommend_jobs(
    "Python SQL Data Science Machine Learning",
    top_n=10
)

display(recommendations)
```

<h2><span style="color:lightblue"><b>Task 3 — forecast monthly posting volume and evaluate</b></span></h2>
A lag-feature Random Forest is evaluated on a chronological holdout, and compared to a naïve last-month baseline. Forecasting is only attempted when enough monthly observations exist.

```python
monthly_total = jobs.dropna(subset=["month"]).groupby("month").size().rename("postings").asfreq("MS", fill_value=0).to_frame()
fig = px.line(monthly_total.reset_index(), x="month", y="postings", markers=True,
              title="Monthly posting volume", color_discrete_sequence=["#7b2cbf"])
fig.show()

def make_lags(s):
    out = s.copy()
    for lag in [1,2,3]:
        out[f"lag_{lag}"] = out["postings"].shift(lag)
    out["rolling_mean_3"] = out["postings"].shift(1).rolling(3).mean()
    out["month_num"] = out.index.month
    out["time_index"] = np.arange(len(out))
    return out.dropna()

fd = make_lags(monthly_total)
if len(fd) >= 8:
    test_n = max(2, min(6, int(np.ceil(len(fd)*.2))))
    train, test = fd.iloc[:-test_n], fd.iloc[-test_n:]
    features = ["lag_1","lag_2","lag_3","rolling_mean_3","month_num","time_index"]
    forecast_model = RandomForestRegressor(n_estimators=300, random_state=42, min_samples_leaf=2)
    forecast_model.fit(train[features], train["postings"])
    pred = forecast_model.predict(test[features])
    naive = test["lag_1"].to_numpy()
    metrics = pd.DataFrame({
        "Model":["Random Forest lag model","Naive last-month baseline"],
        "MAE":[mean_absolute_error(test["postings"],pred), mean_absolute_error(test["postings"],naive)],
        "RMSE":[np.sqrt(mean_squared_error(test["postings"],pred)), np.sqrt(mean_squared_error(test["postings"],naive))],
        "R2":[r2_score(test["postings"],pred), r2_score(test["postings"],naive)]
    })
    display(metrics.round(3))
    compare = pd.DataFrame({"month":test.index, "Actual":test["postings"].values,
                            "Random Forest":pred, "Naive baseline":naive})
    compare = compare.melt("month", var_name="series", value_name="postings")
    fig = px.line(compare, x="month", y="postings", color="series", markers=True,
                  title="Chronological holdout: actual vs predicted monthly postings")
    fig.show()
else:
    print("Not enough monthly observations for a reliable lag-based model; do not report assumed accuracy.")
```

Posting counts increase sharply from January to February and March. This may reflect changes in data collection coverage as well as changes in actual posting activity.

<h2><span style="color:lightblue"><b>Interpretation</b></span></h2>

- The available data indicates a sharp increase in recorded job postings during February and March 2024. However, this should not automatically be interpreted as genuine market growth because the earlier months contain very few records.

<h2><span style="color:lightblue"><b>Task 5 — personalized job recommendation engine</b></span></h2>

TF-IDF and cosine similarity match user-entered interests to titles and inferred categories. The score means text similarity—not hiring probability. Because full job descriptions are unavailable, recommendations are title/category based.



This is the core component of the project.
We can develop a content-based recommendation system using job titles and job categories.

-mHow it works
A user provides their interests, skills, preferred country, and pricing preferences. The system compares those preferences with available job postings and ranks relevant opportunities.

- For example:User profile
- Skills: Python, SQL, Machine Learning
- Preferred category: Data Science
- Preferred country: India
- Pricing preference: Hourly or fixed budget

Recommendation output
The system returns relevant postings with:
- Job title
- Job category
- Country
- Hourly rate or project budget
- Publication date
- Job posting link
- Relevance score

We can use TF-IDF and cosine similarity to calculate textual relevance.

```python
MAX_RECOMMENDER_ROWS = 100_000
rec_jobs = jobs.sample(MAX_RECOMMENDER_ROWS, random_state=42).copy() if len(jobs) > MAX_RECOMMENDER_ROWS else jobs.copy()
rec_jobs["recommendation_text"] = rec_jobs["title_clean"].fillna("") + " " + rec_jobs["job_category"].fillna("")
tfidf = TfidfVectorizer(stop_words="english", ngram_range=(1,2), max_features=30000, min_df=2)
job_matrix = tfidf.fit_transform(rec_jobs["recommendation_text"])

def recommend_jobs(user_skills, top_n=10, country=None, pricing_type=None, min_rate=None):
    if not str(user_skills).strip():
        raise ValueError("Enter at least one skill or job interest.")
    user_vec = tfidf.transform([str(user_skills)])
    scores = cosine_similarity(user_vec, job_matrix).ravel()
    out = rec_jobs.copy()
    out["relevance_score"] = scores
    out = out[out["relevance_score"] > 0]
    if country and country != "All countries":
        out = out[out["country"].str.casefold() == country.casefold()]
    if pricing_type and pricing_type != "All pricing types":
        out = out[out["pricing_type"] == pricing_type]
    if min_rate is not None:
        out = out[out["hourly_midpoint"].ge(min_rate) | out["budget"].ge(min_rate)]
    out = out.sort_values(["relevance_score","published_date"], ascending=[False,False])
    cols = ["title","job_category","country","pricing_type","hourly_low","hourly_high","budget",
            "published_date","link","relevance_score"]
    return out[cols].head(top_n)

display(recommend_jobs("Python SQL data analysis machine learning", top_n=10))
```
